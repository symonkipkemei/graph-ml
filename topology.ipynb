```python
# Import the TopologyGraph class and explore it
from topology.graph import TopologyGraph
import inspect

# Create an instance to explore
cube = TopologyGraph()

print("TopologyGraph instance attributes and methods:")
print("=" * 70)

# Get all public attributes and methods
attrs = [attr for attr in dir(cube) if not attr.startswith('_')]

for attr in attrs:
    obj = getattr(cube, attr)
    if callable(obj):
        # Get method signature
        try:
            sig = inspect.signature(obj)
            print(f"  {attr}{sig}")
        except:
            print(f"  {attr}()")
    else:
        print(f"  {attr} = {type(obj).__name__}")

print("\n" + "=" * 70)
print("Let's check the help for TopologyGraph:")
print("=" * 70)
help(TopologyGraph)
```

```python
# Create a cube topology using TopologyGraph
cube = TopologyGraph()

# Define 8 nodes (vertices of a cube)
nodes = ['node0', 'node1', 'node2', 'node3', 'node4', 'node5', 'node6', 'node7']

# Add nodes to the topology
for node_id in nodes:
    cube.add_node(node_id)

# Define 12 edges (connections between vertices)
edges = [
    # Bottom face
    ('node0', 'node1'),
    ('node1', 'node2'),
    ('node2', 'node3'),
    ('node3', 'node0'),
    # Top face
    ('node4', 'node5'),
    ('node5', 'node6'),
    ('node6', 'node7'),
    ('node7', 'node4'),
    # Vertical edges
    ('node0', 'node4'),
    ('node1', 'node5'),
    ('node2', 'node6'),
    ('node3', 'node7')
]

# Add edges (links) to the topology
for src, dst in edges:
    cube.add_link(src, dst)

print(f"Cube Topology Created!")
print(f"Number of nodes: {len(cube.nodes)}")
print(f"Number of links: {len(cube.links)}")
print(f"\nNodes: {list(cube.nodes.keys())}")
print(f"\nLinks: {list(cube.links.keys())}")
```

```python
# Visualize the cube topology
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define 3D positions for cube vertices
positions = {
    'node0': (0, 0, 0),
    'node1': (1, 0, 0),
    'node2': (1, 1, 0),
    'node3': (0, 1, 0),
    'node4': (0, 0, 1),
    'node5': (1, 0, 1),
    'node6': (1, 1, 1),
    'node7': (0, 1, 1)
}

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Draw edges
for link_id, link in cube.links.items():
    # Get the nodes connected by this link
    nodes_in_link = [n for n in cube.nodes.keys() if n in link_id]
    if len(nodes_in_link) >= 2:
        src, dst = nodes_in_link[0], nodes_in_link[1]
        x = [positions[src][0], positions[dst][0]]
        y = [positions[src][1], positions[dst][1]]
        z = [positions[src][2], positions[dst][2]]
        ax.plot(x, y, z, 'b-', linewidth=2, alpha=0.6)

# Draw nodes
xs = [pos[0] for pos in positions.values()]
ys = [pos[1] for pos in positions.values()]
zs = [pos[2] for pos in positions.values()]
ax.scatter(xs, ys, zs, c='red', s=300, alpha=0.8, edgecolors='black', linewidth=2)

# Add node labels
for node, pos in positions.items():
    ax.text(pos[0], pos[1], pos[2], f'  {node}', fontsize=10, fontweight='bold')

ax.set_xlabel('X', fontsize=12)
ax.set_ylabel('Y', fontsize=12)
ax.set_zlabel('Z', fontsize=12)
ax.set_title('3D Cube Topology (using topology package)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Display detailed topology information
print("Cube Topology Details:")
print("=" * 70)

print("\nNodes:")
for node_id, node in cube.nodes.items():
    print(f"  {node_id}: {node}")

print("\nLinks:")
for link_id, link in cube.links.items():
    print(f"  {link_id}: {link}")
```
