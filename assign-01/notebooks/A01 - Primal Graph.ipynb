## 4. Import the OBJ file

```python
objects = Topology.ByOBJPath(r"E:\softwares-4\graph-ml\assign-01\geometry\box-house.obj", selfMerge=True)
print("Objects is a list")
print(objects)
```

# 5.0 Remove Coplanar surface

```python


for o in objects:
    Topology.RemoveCoplanarFaces(o, 0.01)
    print(f"Object is a {o.Type()}")
```

## 6. Show the geometry

```python
Topology.Show(objects,
              vertexSizeKey="size",
              vertexColorKey="color",
              edgeWidthKey="width",
              edgeColorKey="color",
              faceOpacity=0.1,
              backgroundColor="white",
              width=500,
              height=500,
              renderer=renderer)
```

# Extract CellComplex from the imported cluster

```python

cells = Topology.Cells(objects[0])
print("Number of cells:", len(cells))
cc = CellComplex.ByCells(cells)
print("CellComplex:", cc)
```

# Derive primal graph from the geometry's actual vertices and edges


```python
vertices = Topology.Vertices(cc)
vertices = [Topology.Copy(v) for v in vertices]
edges = Topology.Edges(cc)
edges = [Topology.Copy(e) for e in edges]
g1 = Graph.ByVerticesEdges(vertices, edges)
```

# Assign Visual attributes to edges and vertices

```python
vertices = Graph.Vertices(g1)
for v in vertices:
    d = Dictionary.ByKeysValues(["size", "color"], [18, "red"])
    v = Topology.SetDictionary(v, d)
edges = Graph.Edges(g1)
for e in edges:
    d = Dictionary.ByKeysValues(["width", "color"], [4, "black"])
    e = Topology.SetDictionary(e, d)
```

# Show Graph

```python
Topology.Show(g1,
              vertexSizeKey="size",
              vertexColorKey="color",
              edgeWidthKey="width",
              edgeColorKey="color",
              faceOpacity=0.3,
              backgroundColor="white",
              width=500,
              height=500,
              renderer=renderer)
```

# Show Dual Graph with Geometry

```python
Topology.Show(cc, g1,
              vertexSizeKey="size",
              vertexColorKey="color",
              edgeWidthKey="width",
              edgeColorKey="color",
              faceOpacity=0.3,
              backgroundColor="white",
              width=500,
              height=500,
              renderer=renderer)
```
